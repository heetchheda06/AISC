```python
import numpy as np
import matplotlib.pyplot as plt

N = 1000
bits = np.random.randint(0, 2, N)

bpsk_signal = 2 * bits - 1

SNR_dB = 5
SNR = 10 ** (SNR_dB / 10)

noise_variance = 1 / (2 * SNR)
noise = np.sqrt(noise_variance) * np.random.randn(N)

received_signal = bpsk_signal + noise

received_bits = (received_signal > 0).astype(int)

errors = np.sum(bits != received_bits)
BER = errors / N

print("Number of Errors:", errors)
print("Bit Error Rate (BER):", BER)

plt.figure(figsize=(10, 3))
plt.stem(bits[:20])
plt.title("Original Bits")
plt.xlabel("Bit Index")
plt.ylabel("Bit Value")
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 3))
plt.plot(bpsk_signal[:50], marker='o')
plt.title("BPSK Modulated Signal")
plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 3))
plt.plot(received_signal[:50], marker='o')
plt.title("Received Signal with AWGN")
plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```

Multipath fading

```python
import math
import matplotlib.pyplot as plt

distance = [1, 2, 4, 6, 8, 10]
PL_d0 = 40
n = 3

path_loss = []

for d in distance:
    PL = PL_d0 + 10 * n * math.log10(d / 1)
    path_loss.append(PL)

print("Distance (m)   Path Loss (dB)")
print("-----------------------------")

for i in range(len(distance)):
    print(distance[i], "        ", round(path_loss[i], 2))

plt.plot(distance, path_loss, marker="o")
plt.xlabel("Distance (m)")
plt.ylabel("Path Loss (dB)")
plt.title("Indoor Path Loss vs Distance")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 100, 101)

BS1 = 0
BS2 = 100

signal_BS1 = -50 - 0.5 * x
signal_BS2 = -100 + 0.5 * x

handover_point = 50

plt.figure(figsize=(8, 5))
plt.plot(x, signal_BS1, label="BS1 Signal")
plt.plot(x, signal_BS2, label="BS2 Signal")
plt.axvline(handover_point, linestyle="--", label="Handover Point")
plt.scatter(handover_point, -75, s=80)
plt.xlabel("Mobile Position (m)")
plt.ylabel("Signal strength (dBm)")
plt.title("Handover Behaviour While Moving")
plt.grid(True)
plt.legend()
plt.show()

print("Mobile starts connected to BS1")
print("Mobile moves from BS1 towards BS2")
print("Signal from BS1 decreases")
print("Signal from BS2 increases")
print("Handover occurs at approximately 50 m")
print("Mobile is now connected to BS2")
```
